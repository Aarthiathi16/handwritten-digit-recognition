```python
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# Sample movie dataset
data = {
    'Movie': [
        'Avengers', 'Iron Man', 'Batman',
        'Titanic', 'The Notebook', 'Interstellar',
        'Inception', 'Joker', 'Avatar'
    ],
    'Genre': [
        'Action Superhero', 'Action Superhero',
        'Action Crime', 'Romance Drama',
        'Romance Drama', 'Sci-Fi Adventure',
        'Sci-Fi Action', 'Crime Drama',
        'Sci-Fi Adventure'
    ]
}

df = pd.DataFrame(data)

# Convert genres into numerical features
cv = CountVectorizer()
genre_matrix = cv.fit_transform(df['Genre'])

# Calculate similarity between movies
similarity = cosine_similarity(genre_matrix)

# Function to recommend movies
def recommend_movies(movie_name):
    if movie_name not in df['Movie'].values:
        print("Movie not found in the database!")
        return

    index = df[df['Movie'] == movie_name].index[0]

    # Get similarity scores
    scores = list(enumerate(similarity[index]))

    # Sort movies by similarity
    scores = sorted(scores, key=lambda x: x[1], reverse=True)

    print("\nMovies recommended for", movie_name, ":")

    count = 0
    for i, score in scores:
        if i != index and score > 0:
            print(df.iloc[i]['Movie'])
            count += 1

            if count == 3:
                break

    if count == 0:
        print("No similar movies found.")

# Get user input
movie = input("Enter a movie name: ")
recommend_movies(movie)
```
